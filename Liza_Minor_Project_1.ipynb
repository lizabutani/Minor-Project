MULTIVARIATE ANALYSIS

```python
# Clean numeric columns to get a meaningful correlation matrix
df['milage_num'] = df['milage'].astype(str).str.replace(' mi.', '').str.replace(',', '').astype(float)
df['price_num'] = df['price'].astype(str).str.replace('$', '').str.replace(',', '').astype(float)

# Compute the correlation matrix for all numeric fields
correlation_matrix = df[['model_year', 'milage_num', 'price_num']].corr()

# Plot the correlation heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Multivariate Feature Correlation Heatmap")
plt.show()
```

UNIVARIATE ANALYSIS

```python
#OUTLIERS
plt.figure(figsize=(10, 5))
sns.boxplot(x=df["price"])
plt.title("PRICE OUTLIER DETECTION")
plt.xlabel("Price ($)")
plt.show()
```

BIVARIATE ANALYSIS

```python
#CATEGORICAL VS CATEGORICAL
plt.figure(figsize=(12, 6))
sns.countplot(x="fuel_type", hue="accident", data=df)
plt.title("ACCIDENT PROFILE BY FUEL TYPE")
plt.show()
```

```python
#CATEGORICAL VS NUMERICAL
plt.figure(figsize=(10, 6))
sns.boxplot(x="accident", y="price", data=df)
plt.title("CAR PRICE DISTRIBUTION BY ACCIDENT PROFILE")
plt.show()
```

INSIGHTS

```python
print("Insight 1: Average Price by Accident Profile")
print(df.groupby('accident')['price_num'].mean())

print("\nInsight 2: Fuel Type Market Volume")
print(df['fuel_type'].value_counts(dropna=False))

print("\nInsight 3: Clean Title Blank Field Check")
print(df['clean_title'].isnull().sum())

print("\nInsight 4: Average Mileage by Age (Model Year)")
df['car_age'] = 2026 - df['model_year']
print(df.groupby('car_age')['milage_num'].mean().head(10))

print("\nInsight 5: Top 5 Most Common Brands")
print(df['brand'].value_counts().head(5))
```

OTHER CHARTS FOR ANALYSIS


```python
plt.figure(figsize=(10, 5))
sns.histplot(df['price_num'], kde=True, bins=50, color="red")
plt.title("CAR PRICE DISTRIBUTION")
plt.xlabel("Price ($)")
plt.ylabel("Count")
plt.show()
```

Histogram Visually captures that  car prices are heavily right-skewed, showing most inventory.

```python
plt.figure(figsize=(10, 5))
sns.boxplot(x=df["milage_num"], color="blue")
plt.title("VEHICLE MILEAGE OUTLIER DETECTION")
plt.xlabel("Mileage (mi)")
plt.show()
```
